```python
import os

BATCH_SIZE = 1000
OUTPUT_DIR = "mtgjson_card_batches"

# Create output directory
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Download AllPrintings.json
url = "https://mtgjson.com/api/v5/AllPrintings.json"
response = requests.get(url)

if response.status_code == 200:
    data = response.json()["data"]

    cards_collected = []
    batch_number = 1

    for set_code, set_info in data.items():
        for card in set_info.get("cards", []):
            cards_collected.append(card)
            if len(cards_collected) == BATCH_SIZE:
                # Write this batch to a file
                filename = os.path.join(OUTPUT_DIR, f"cards_batch_{batch_number:03}.json")
                with open(filename, "w") as f:
                    json.dump(cards_collected, f, indent=2)
                print(f"Saved {filename} with {len(cards_collected)} cards")
                cards_collected = []
                batch_number += 1

    # Save remaining cards
    if cards_collected:
        filename = os.path.join(OUTPUT_DIR, f"cards_batch_{batch_number:03}.json")
        with open(filename, "w") as f:
            json.dump(cards_collected, f, indent=2)
        print(f"Saved final {filename} with {len(cards_collected)} cards")
```

```python
def extract_card_info(card):
    legalities = card.get("legalities", {})
    legal_in = [fmt for fmt, status in legalities.items() if status == "Legal"]

    return {
        "uuid": card.get("uuid"),
        "name": card.get("name"),
        "type": card.get("type"),
        "types": card.get("types", []),
        "subtypes": card.get("subtypes", []),
        "manaCost": card.get("manaCost"),
        "convertedManaCost": card.get("convertedManaCost"),
        "colors": card.get("colors", []),
        "colorIdentity": card.get("colorIdentity", []),
        "text": card.get("text"),
        "power": card.get("power"),
        "toughness": card.get("toughness"),
        "loyalty": card.get("loyalty"),
        "rarity": card.get("rarity"),
        "isLegendary": card.get("isLegendary", False),
        "layout": card.get("layout"),
        "setCode": card.get("setCode"),
        "legalities": legal_in
    }
```

```python
import polars as pl

def load_all_cards_from_folder(folder_path):
    data = []

    for file_name in os.listdir(folder_path):
        if file_name.endswith(".json"):
            with open(os.path.join(folder_path, file_name), "r", encoding="utf-8") as f:
                try:
                    cards = json.load(f)
                    if isinstance(cards, dict):
                        # Handles a JSON file with one card
                        data.append(extract_card_info(cards))
                    elif isinstance(cards, list):
                        for card in cards:
                            data.append(extract_card_info(card))
                except Exception as e:
                    print(f"Failed to load {file_name}: {e}")

    return pl.DataFrame(data)
```

```python
url = "https://mtgjson.com/api/v5/AllPricesToday.json"
response = requests.get(url)

BATCH_SIZE = 1000
OUTPUT_DIR = "mtgjson_card_price_batches"

# Create output directory
os.makedirs(OUTPUT_DIR, exist_ok=True)

if response.status_code == 200:
    price_data = response.json()['data']
```

```python
def extract_paper_prices(uuid, price_data):
    row = {'uuid':uuid}
    paper_price = price_data.get("paper", {})

    for vendor, vendor_price in paper_price.items():
        retail_price = vendor_price.get("retail", {})

        for version in ["normal", "foil"]:
            version_prices = retail_price.get(version, {})
            if version_prices:
                latest_date = max(version_prices.keys())
                col_name = f"{vendor}_{version}"
                row[col_name] = version_prices[latest_date]
    return row
```

```python
rows = []
for uuid, card_data in price_data.items():
    rows.append(extract_paper_prices(uuid, card_data))
```

```python
# Ensure all rows have consistent schemas by filling missing values with None
rows = [extract_paper_prices(uuid, price_data[uuid]) for uuid in price_data]

# Convert rows to a DataFrame, ensuring consistent schema
card_price_df = pl.DataFrame(rows, infer_schema_length=None).fill_null(0)

card_price_df.head()
```

```python
legal_price_df = commander_legal_cards_df.join(card_price_df, on="uuid", how="inner")
```

```python
def card_finder(name: str, dataframe: pl.DataFrame):
    return dataframe.filter(pl.col('name') == f'{name}')
```

```python
legal_price_df = legal_price_df.with_columns(
    mean=pl.mean_horizontal(
        "cardmarket_normal",
        "cardsphere_normal",
        "tcgplayer_normal",
        "cardkingdom_normal",
        ignore_nulls=True
    )
)
legal_price_df
```

```python
from sklearn.preprocessing import LabelEncoder

rarity_le = LabelEncoder()

rarity_encoded = rarity_le.fit_transform(legal_price_df["rarity"])
```

```python
legal_price_df = legal_price_df.with_columns([
    pl.Series(name="rarity_encoded", values=rarity_encoded)
])
commander_legal_price_df = legal_price_df.drop("legalities")
commander_legal_price_df.head()
```
